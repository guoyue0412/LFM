```python
# apply the style
set_style_params()

ratio = 0.75
# define the figsize
# width = 3.15 # 0.5 page
# width = 4.33 # 2/3 page
width = 6.69 # full page


# Create the figure and 3D axis
fig = plt.figure(figsize=(width, width * ratio))
ax = fig.add_subplot(111, projection='3d')

# Grid size for the planes
x_label = -1*np.array([0, 0.5, 1.0, 1.5])  # X coordinates
y_label = np.array([0.6, 0.8, 0.9])     # Y coordinates
z_label = np.array([0, 0.8, 0.9, 1.0])     # Z coordinates


# Create xy-plane
# x = np.linspace(-1.5, 1.5, 1000) 
# y = np.linspace(0, 2, 1000) 
# X, Y = np.meshgrid(x, y) 
# for z_v in z_label:
#     Z = np.zeros_like(X) + z_v
#     # ax.plot_surface(X, Y, Z, alpha=0.3, color='gray', edgecolor='none')
#     ax.plot_surface(X, Y, Z, alpha=0.3, color='blue', edgecolor='none')

# Create zy-plane
z = np.linspace(0., 2., 1000) 
y = np.linspace(0., 2., 1000) 
Z, Y = np.meshgrid(z, y) 
for x_v in x_label:
    X = np.zeros_like(Z) + x_v
    # ax.plot_surface(X, Y, Z, alpha=0.1, color='gray', edgecolor='none')
    ax.plot_surface(X, Y, Z, alpha=0.1, color='blue', edgecolor='none')

# Create zx-plane
z = np.linspace(0., 2., 1000) 
x = np.linspace(-1.5, 1.5, 1000) 
Z, X = np.meshgrid(z, x) 
for y_v in y_label:
    Y = np.zeros_like(Z) + y_v
    ax.plot_surface(X, Y, Z, alpha=0.5, color='gray', edgecolor='none')
    # ax.plot_surface(X, Y, Z, alpha=0.2, color='red', edgecolor='none')

# set the axis label
ax.set_xlabel('X/D')
ax.set_ylabel('Y/R')
ax.set_zlabel('Z/H')

# hide the axis
ax.set_axis_off()
# hide the grid
ax.grid(False)

# adjust the view angle
ax.view_init(elev=15, azim=20)
# save the figure
figure_name = 'output/3D_illustration.png'
plt.savefig(figure_name, 
            format='png', 
            bbox_inches='tight', 
            dpi=600
            )
```
